```python
df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D', 'E']
})

df['SerialNo'] = np.arange(5)

print(df)
```

```python
df.drop('SerialNo', axis=1, inplace=True)

print(df)
```

```python
df.drop('SerialNo', axis=1, inplace=True, errors='ignore')

print(df)
```

```python
data = {
    'Movie': ['3 Idiots', 'Dangal', 'PK'],
    'Year': [2009, 2016, 2014],
    'Rating': [8.4, 8.3, 8.1]
}

movie_df = pd.DataFrame(data)
print(movie_df)
```

```python
new_movie_df = movie_df[['Movie', 'Rating', 'Director']]
```
