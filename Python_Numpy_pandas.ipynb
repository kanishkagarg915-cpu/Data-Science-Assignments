```python
import numpy as np

np.random.seed(42)

marks = np.random.randint(20, 101, size=(12, 4))

print("Marks:")
print(marks)

student_avg = np.mean(marks, axis=1)
test_mean = np.mean(marks, axis=0)
test_std = np.std(marks, axis=0)

passed_all = np.all(marks >= 40, axis=1)
high_avg = np.where(student_avg >= 75)[0] + 1
pass_count = np.sum(passed_all)

most_variable_test = np.argmax(test_std) + 1

print("\nStudent averages:")
print(np.round(student_avg, 2))

print("\nTest means:")
print(np.round(test_mean, 2))
print("\nTest standard deviations:")
print(np.round(test_std, 2))

print("\nStudents passing all tests:", pass_count)
print("Students with average >= 75:", high_avg)
print("Test with greatest variation:", most_variable_test)
```

```python
import numpy as np
import pandas as pd

arr = np.arange(21, 46).reshape(5, 5)

print("Original matrix:")
print(arr)

central = arr[1:4, 1:4]
print("\nCentral 3x3 matrix:")
print(central)

transposed = central.T
print("\nTransposed matrix:")
print(transposed)

product = central @ transposed
print("\nMatrix product:")
print(product)

row_sums = np.sum(product, axis=1)
print("\nRow-wise sums:")
print(row_sums)

maximum = np.max(row_sums)
positions = np.where(row_sums == maximum)[0]
print("Maximum row sum:", maximum)
print("Positions (zero-based):", positions)
print("Rows (one-based):", positions + 1)
```

```python
import numpy as np

quantities = np.array([
    [20, 15, 30, 10],
    [25, 18, 20, 12],
    [15, 22, 25, 20],
    [30, 16, 18, 15],
    [22, 25, 15, 18],
    [28, 20, 22, 25]
])

prices = np.array([150, 240, 90, 320])
daily_revenue = quantities @ prices


product_quantity = np.sum(quantities, axis=0)


product_revenue = product_quantity * prices
highest_day = np.argmax(daily_revenue) + 1


best_product = np.argmax(product_quantity) + 1
new_prices = prices.copy()
new_prices[product_quantity > 100] *= 0.90
new_total_revenue = np.sum(quantities @ new_prices)

print("Daily revenue:", daily_revenue)
print("Product quantities:", product_quantity)
print("Product revenue:", product_revenue)
print("Highest-revenue day:", highest_day)
print("Best-selling product:", best_product)
print("Original prices:", prices)
print("Revised prices:", new_prices)
print("Revised total revenue:", new_total_revenue)
```

```python
import numpy as np

np.random.seed(42)

data = np.random.normal(50, 12, 100)

outside_count = np.sum((data < 30) | (data > 70))

cleaned = np.where(
    data < 30,
    30,
    np.where(data > 70, 70, data)
)

print("Values outside 30-70:", outside_count)

print("\nOriginal statistics:")
print("Mean:", np.mean(data))
print("Variance:", np.var(data))
print("Standard deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))

print("\nCleaned statistics:")
print("Mean:", np.mean(cleaned))
print("Variance:", np.var(cleaned))
print("Standard deviation:", np.std(cleaned))
print("Minimum:", np.min(cleaned))
print("Maximum:", np.max(cleaned))
```

```python
import pandas as pd

branch_a = pd.Series({"Q1": 120, "Q2": 145, "Q4": 180})
branch_b = pd.Series({"Q2": 110, "Q3": 135, "Q4": 155})
branch_c = pd.Series({"Q1": 95, "Q3": 125, "Q4": 140})

# Ordinary addition
ordinary = branch_a + branch_b + branch_c

print("Ordinary addition:")
print(ordinary)

# Quarters common to all branches
common = (
    branch_a.index
    .intersection(branch_b.index)
    .intersection(branch_c.index)
)

print("\nCommon quarters:", common.tolist())

combined_common = (
    branch_a[common] +
    branch_b[common] +
    branch_c[common]
)

print("\nCombined earnings for common quarters:")
print(combined_common)

# Treat missing branch records as zero
combined_zero = (
    branch_a.add(branch_b, fill_value=0)
            .add(branch_c, fill_value=0)
)

print("\nCombined earnings treating missing records as zero:")
print(combined_zero)
```

```python
import pandas as pd

scores = {
    "R101": 78,
    "R102": 65,
    "R104": 92,
    "R105": 48
}

required = ["R105", "R103", "R101", "R106", "R104", "R102"]

s = pd.Series(scores, index=required)

print("Original Series:")
print(s)

print("\nMissing student records:")
print(s[s.isnull()])

print("\nValid scores >= 70:")
valid = s[s.notnull() & (s >= 70)]
print(valid)

# Convert to dictionary
data = valid.to_dict()
print("\nDictionary:")
print(data)
new_series = pd.Series(data)

print("\nReconstructed Series:")
print(new_series)
print("\nVerification:")
print(valid.equals(new_series))
```
